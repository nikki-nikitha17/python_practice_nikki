```python
brand_avg_prices = df.groupby("Brand")["Price"].mean().sort_values(ascending=False)
brand_avg_prices.count()
brand_avg_prices
```

```python
d={}
for i in brand_avg_prices.index:
  print(i)
```

```python
d = {}
j=len(brand_avg_prices)
for i in brand_avg_prices.index:
  d[i] = j
  j-=1
d
```

```python
x=[]
for j in df['Brand']:
  y=d[j]
  x.append(y)
df['encoded_brand']=x
df
```

```python
scaled_fuel_efficieny=[]
min=df['Fuel Efficiency'].min()
max=df['Fuel Efficiency'].max()
for i in df['Fuel Efficiency']:
  scaled_value=(i-min)/(max-min)
  scaled_fuel_efficieny.append(scaled_value)
df['scaled_fuel_efficieny']=scaled_fuel_efficieny
df
```

```python
scaled_horsepower=[]
min=df['Horsepower'].min()
max=df['Horsepower'].max()
for i in df['Horsepower']:
  scaled_value=(i-min)/(max-min)
  scaled_horsepower.append(scaled_value)
df['scaled_horsepower']=scaled_horsepower
df
```

```python
scaled_encoded_brand=[]
min=df['encoded_brand'].min()
max=df['encoded_brand'].max()
for i in df['encoded_brand']:
  scaled_value=(i-min)/(max-min)
  scaled_encoded_brand.append(scaled_value)
df['scaled_encoded_brand']=scaled_encoded_brand
df
```

```python
df['Performance_Index'] = (df['Horsepower'] * 0.7 + df['Fuel Efficiency'] * 0.3) / df['Price']
df
```
